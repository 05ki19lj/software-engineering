```python
# ================================
# Colab 全自動化初始化 + VOT/PySOT + CSRT 跟踪 + MP4 輸出
# ================================

import os
from google.colab import drive
from IPython.display import display, HTML
import cv2
import numpy as np

# ------------------------------
# 0️⃣ 掛載 Google Drive
# ------------------------------
drive.mount('/content/drive', force_remount=True)

project_path = '/content/drive/MyDrive/000/video_analyst'
os.chdir(project_path)
print("當前資料夾:", os.getcwd())
!ls

# ------------------------------
# 1️⃣ 安裝與升級套件
# ------------------------------
!pip install --quiet stack_data opencv-python matplotlib
!pip install --quiet --upgrade "Cython>=0.29.36" "scipy>=1.10.0" "numba>=0.58.0"

# 產生暫時 requirements
!sed -E 's/Cython==0.27.0/Cython>=0.29.36/; \
        s/scipy==1.4.1/scipy>=1.10.0/; \
        s/numba==0.45.0/numba>=0.58.0/' requirements.txt > temp_requirements.txt

# 安裝其他套件
!pip install --quiet -r temp_requirements.txt --upgrade --no-deps

# ------------------------------
# 2️⃣ 編譯 VOT/PySOT 工具
# ------------------------------
vot_utils_path = 'videoanalyst/evaluation/vot_benchmark/pysot/utils'
if os.path.exists(vot_utils_path):
    os.chdir(vot_utils_path)
    !python3 setup.py build_ext --inplace
    os.chdir(project_path)
    print("VOT/PySOT 工具編譯完成")
else:
    print("找不到 VOT/PySOT 工具資料夾，請確認路徑")

# ------------------------------
# 3️⃣ 顯示 GPU 資訊
# ------------------------------
!nvidia-smi

# ------------------------------
# 4️⃣ 下載範例影片
# ------------------------------
!wget -O /content/sample_video.avi https://github.com/opencv/opencv/raw/master/samples/data/vtest.avi
video_path = '/content/sample_video.avi'
output_path = '/content/drive/MyDrive/000/result.mp4'  # 改成 mp4

# ------------------------------
# 5️⃣ CSRT 跟踪
# ------------------------------
cap = cv2.VideoCapture(video_path)
ret, frame = cap.read()
if not ret:
    raise Exception("影片無法讀取")

# 初始化 CSRT 跟踪器
tracker = cv2.TrackerCSRT_create()
bbox = (200, 50, 100, 80)  # 初始跟踪框
tracker.init(frame, bbox)

# 輸出影片設定 (mp4 + H.264)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 'mp4v' H.264
out = cv2.VideoWriter(output_path, fourcc, 20.0, (frame.shape[1], frame.shape[0]))

# 逐幀處理
while True:
    ret, frame = cap.read()
    if not ret:
        break
    success, bbox = tracker.update(frame)
    if success:
        x, y, w, h = map(int, bbox)
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
    out.write(frame)

cap.release()
out.release()

# ------------------------------
# 6️⃣ 在 Colab 播放結果影片
# ------------------------------
display(HTML(f"""
<video width="640" height="480" controls>
  <source src="{output_path}" type="video/mp4">
  您的瀏覽器不支援影片播放。
</video>
"""))

print("影片處理完成，已輸出至:", output_path)
```
